```python
class FinancialAccount:
    def __init__(self, fc_account, fc_balance):
        self.account = fc_account
        self.balance = fc_balance
        self.history = []
        self.settings = {"limit": 100_000, "currency": "UAN"}

    def top_up(self, amount):
        self.balance += amount
        self.history.append(f"Поповнення на суму: {amount}.")

    def withdrawl(self, amount):
        self.balance -= amount
        self.history.append(f"Зняття на суму: {amount}.")

    def display_info(self):
        print(f"Номер рахунку: {self.account}")
        print(f"Поточний баланс: {self.balance} {self.settings['currency']}")
        print(f"Історія транзакцій: {self.history}")
```

```python
class DepositAccount(FinancialAccount):
    def __init__(self, fc_account, fc_balance, fc_rate, fc_name, fc_surname, fc_phone):
        super().__init__(fc_account, fc_balance)
        self.rate = fc_rate
        self.client_profile = Client(fc_name, fc_surname, fc_phone)

    def calculate_interest(self):
        profit = self.balance * self.rate
        self.balance += profit
        self.history.append(f"нарахування відсотків з депозиту на суму: {profit}.")

    def activate_deposit(self):
        self.client_profile.verify_identity()

    def display_info(self):
        print(f"Інформація про депозит клієнта {self.client_profile.name}")
        super().display_info()
        print("Відсоткова ставка: {self.rate * 100}%")
```

```python
class Client:
    def __init__(self, fc_name, fc_surname, fc_phone):
        self.name = fc_name
        self.surname = fc_surname
        self.phone = fc_phone
        self.is_verified = False

    def verify_identity(self):
        self.is_verified = True
        print(f"Особу клієнта {self.name} {self.surname} успішно підтверджено!")
```

```python
my_deposit = DepositAccount("UA123456789", 10000, 0.05, "Анастасія", "Гарбар", "+38012435464")
my_deposit.activate_deposit()

my_deposit.top_up(5_000)
my_deposit.withdrawl(2_000)

my_deposit.calculate_interest()

my_deposit.display_info()
```
